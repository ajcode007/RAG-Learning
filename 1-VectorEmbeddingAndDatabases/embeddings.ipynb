## Create Your First Embeddings

```python
### Huggingface And OpenAI Models

from langchain_huggingface import HuggingFaceEmbeddings

## Initialize a simple Embedding Model (no API key needed)

embeddings = HuggingFaceEmbeddings(
    model_name = "sentence-transformers/all-MiniLM-L6-v2"
)
embeddings
```

```python
## Create your first embeddings
text = "Hello, I am learning about RAG"

embedding = embeddings.embed_query(text)
print(f"Text: {text}")
print(f"Embedding length: {len(embedding)}")
print(embedding)
```

```python
sentences = [

"what was YOUR specific task in it? (not team's job — ur job)",

"what tools/method did u use? (sql, pyspark, excel, hadoop, dashboard tool, etc)",

"who consumed ur output? (leadership? another team? automated system?)",

"what changed because of ur work? (time saved, error reduced, decision enabled, accuracy improved — even rough estimate ok, exact number not needed yet)",

"any number u remember? (rows processed, % improvement, hours saved, frequency — daily/weekly/monthly)"

]

embedding_sentence = embeddings.embed_documents(sentences)
print(embedding_sentence[0])
```
